### 1.3 Sample Dataset

````python
emp_df = pd.DataFrame(
    {
        "Name": ["Alex", "Ajax", "Jane", "John", "Karen"],
        "Role": ["Senior SD", "Junior SD", "Junior SD", "Architect", "As. Architect"],
        "TotExp": [9, 6, 3, 15, 10],
        "Python": [4, 1, 2, 5, 4],
        "MongoDB": [1, 3, None, 5, 5],
        "MySQL": [3, 2, 2, 6, None],
        "JavaScript": [None, 3, None, 1, 1],
    }
)

row2, col2 = emp_df.shape
print(f"DataFrame has {row2} rows and {col2} columns.")

emp_df
````

````python
emp_df.info()
````

## 2 Melting

Converting data from wide format to long format.

### 2.1 Example

##### Original DataFrame

````python
emp_df
````

##### Melted DataFrame

````python
emp_long_df = emp_df.melt(
    id_vars="Name",
    value_vars=["Python", "MongoDB", "MySQL", "JavaScript"],
    var_name="Tech",
    value_name="Exp",
)

row3, col3 = emp_long_df.shape
print(f"DataFrame has {row3} rows and {col3} columns.")

emp_long_df
````

## 3 Pivoting

Converting data from long format to wide format.

### 3.1 Example

##### Original DataFrame

````python
emp_df.sort_values(by="Name")
````

##### Pivoted DataFrame

If we pivot the melted DataFrame we almost bet back original DataFrame.

````python
emp_wide_df = emp_long_df.pivot(columns="Tech", index="Name", values="Exp")

row4, col4 = emp_wide_df.shape
print(f"DataFrame has {row4} rows and {col4} columns.")

emp_wide_df
````

````python
emp_wide_df.reset_index().sort_values(by="Name")
````

## 4 Binning

### 4.1 Example

##### Original DataFrame

````python
emp_df
````

````python
exp_in_years = [0, 5, 10, 12, np.inf]
roles = [
    "Junior SD",  # (0, 5]
    "Senior SD",  # (5, 10]
    "As. Architect",  # (10, 13]
    "Architect",  # (12, Inf]
]

# Create new column `ElgRole` for eligible role.
emp_df["ElgRole"] = pd.cut(emp_df["TotExp"], bins=exp_in_years, labels=roles)
````

##### After Binning

````python
emp_df[["Name", "TotExp", "Role", "ElgRole"]].sort_values(by="TotExp")
````

## 5 Handling missing values

### 5.1 `null` representation

Various data types and their null representation.

<div style="display: inline-block">

| Data Type | Null representation |
| --------- | ------------------- |
| String    | None                |
| Float     | NaN                 |
| Boolean   | \<NA\>              |
| DateTime  | NaT                 |

</div>

> **Note**:
>
> Integer datatype does not have any null representation. Integer values are converted to Float and `NaN` represents their null value.

````python
from datetime import datetime as dt

dtypes_df = pd.DataFrame(
    {
        "String": ["A", None, "C", "D", "E"],
        "Integer": [1, 2, 3, None, 5],
        "Float": [None, 2.7, 3.14, 4.0, 5],
        "Boolean": [True, False, None, True, True],
        "DateTime": [dt.now(), dt(2020, 5, 17), None, None, dt(2025, 7, 12)],
        "NoNulls": ["A", 2, 3.14, True, datetime.now()],
    }
)

# dtypes_df["Integer"] = dtypes_df["Integer"].astype("int64")
dtypes_df["Boolean"] = dtypes_df["Boolean"].astype("boolean")

row_count, col_count = dtypes_df.shape

display(dtypes_df)
print(f"Dataframe contains {row_count} rows and {col_count} columns.\n\n")

dtypes_df.info()
````
