## 3 Concatenation

### 3.1 Row Concatenation

#### Insert single row

```python
# New employee record.
new_emp = [
    {
        "Name": "Bob",
        "Age": 30,
        "Role": "Junior SD",
        "D.O.J": "01-07-2024",
    }
]

# Create new DataFrame for new employee record.
new_emp_df1 = pd.DataFrame(data=new_emp)
new_emp_df1
```

Observe that new DataFrame has explicit index `0`

```python
# Concatenate new DataFrame to main DataFrame.
emp_df = pd.concat([emp_df, new_emp_df1])
emp_df
```

> **Note**:
>
> If `ignore_index` parameter of `concat` method is not set to `True` then explicit indices from both the DataFrames are retained as it is.

#### Insert multiple rows

```python
new_emps = [
    ["David", 30, "Junior SD", "01-07-2024"],
    ["Karen", 33, "Associate Architect", "01-07-2024"],
]
column_names = ["Name", "Age", "Role", "D.O.J"]

new_emp_df2 = pd.DataFrame(data=new_emps, columns=column_names)
new_emp_df2
```

```python
# Concatenate new DataFrame to main DataFrame.
emp_df = pd.concat([emp_df, new_emp_df2], axis=0, ignore_index=True)
emp_df
```

Check explicit indices

```python
emp_df.index.values
```

Check implicit indices

```python
np.where(emp_df.index.values)
[emp_df.index.get_loc(exp_idx) for exp_idx in emp_df.index.values]
```

> **Note**:
>
> `concat()` method does not have `inplace` parameter.

## 4 Merging

Merging is done using Joins

```python
skill_data = [
    ["Ajax", "Python"],
    ["Alex", "Python"],
    ["Alex", "MySQL"],
    ["Ajax", "MongoDB"],
    ["John", "HLD+LLD"],
    ["Anna", "Business Administration"],
    ["", "JAVA"],
]

skill_df = pd.DataFrame(data=skill_data, columns=["Name", "Skill"])
```

```python
skill_df.head()
```

### 4.1 Inner Join

`how` by default is set to `inner`

```python
pd.merge(emp_df, skill_df)
```

Rename "Name" column in `skill_df` to "EmpName" for clarity.

```python
skill_df.rename({"Name": "EmpName"}, axis=1, inplace=True)
```

```python
emp_df.merge(skill_df, left_on="Name", right_on="EmpName", how="inner")
```

### 4.2 Left Outer Join

```python
emp_df.merge(skill_df, left_on="Name", right_on="EmpName", how="left")
```

### 4.3 Right Outer Join

```python
emp_df.merge(skill_df, left_on="Name", right_on="EmpName", how="right")
```

### 4.4 Full Outer Join

```python
emp_df.merge(skill_df, left_on="Name", right_on="EmpName", how="outer")
```

When the column names are same pandas adds suffixes `_x`, `_y` to the column names. For example:

```python
df1 = pd.DataFrame(
    {
        "A": [10, 30],
        "B": [20, 40],
        "C": [30, 60],
    }
)
df2 = pd.DataFrame(
    {
        "A": [10, 30],
        "C": [30, 60],
    }
)
df2.merge(df1, on="A", how="outer")
```
