## 2 Shallow vs Deep Copy

### 2.1 Python list

#### Shallow Copy in Python List

```python
a = [1, 2, 3, 4]
b = a

a[1] = 100
b
```

```python
a is b
```

#### Deep Copy in Python List

```python
a = [1, 2, 3, 4]
b = a[:]

a[1] = 100
b
```

```python
a is b
```

### 2.2 NumPy array

#### Shallow Copy in NumPy array

```python
a = np.array([1, 2, 3, 4])
b = a.reshape(2, 2)

a[1] = 100
b
```

```python
a is b
```

```python
np.shares_memory(a, b)
```

#### Deep Copy in NumPy array

```python
a = np.array([1, 2, 3, 4])
b = a + 0

a[1] = 100
b
```

```python
a is b
```

```python
np.shares_memory(a, b)
```

### 2.3 `np.view` vs `np.copy`

#### `np.view()`

##### Example #1

```python
a = np.array([1, 2, 3, 4])
b = a.view()

a[1] = 100
b
```

```python
a is b
```

```python
np.shares_memory(a, b)
```

##### Example #2

```python
a = np.array([1, 2, 3, 4])
b = (a + 0).view()

a[1] = 100
b
```

```python
a is b
```

```python
np.shares_memory(a, b)
```

#### `np.copy()`

##### Example #1

```python
a = np.array([1, 2, 3, 4])
b = a.copy()

a[1] = 100
b
```

```python
a is b
```

```python
np.shares_memory(a, b)
```

```python
a = np.array([0, 1, 2, 3, 4, 5])
b = a[a % 1 == 0]
b[0] = 10
a[:2]
```

## 3 Manipulation routines

### 3.1 Spliting

#### `np.split`

##### Example #1: One dimensional array

```python
arr = np.arange(0, 8)
arr
```

Default value for `axis` parameter of `np.split()` is `0`.

```python
np.split(arr, 2)
```

```python
try:
    np.split(arr, 3)

except ValueError as err:
    print("ValueError:", err)
```

```python
splits = np.split(arr, indices_or_sections=[2, 5, 7, 8, 100])
splits
```

> **Note**:
>
> `indices_or_sections` accepts values out of range

```python
type(splits)
```

```python
np.split(arr, 2, axis=0)
```

##### Example #2: Two dimensional array

```python
arr = np.arange(1, 17).reshape(4, 4)
arr
```

Default value for `axis` parameter is `0`.

```python
np.split(arr, 2)
```

When `axis` is explicitly set to `0`.

```python
np.split(arr, 2, axis=0)
```

When `axis` is explicitly set to `1`.

```python
np.split(arr, 2, axis=1)
```

### 3.2 Stacking

#### `np.concatenate`

##### Example #1: One dimensional array

```python
arr1 = np.arange(0, 4)
arr1
```

```python
arr2 = np.arange(5, 9)
arr2
```

Default value for `axis` parameter of `np.concatenate()` is `0`.

```python
np.concatenate((arr1, arr2))
```

```python
np.concatenate((arr1, arr2), axis=0)
```

```python
try:
    np.concatenate((arr1, arr2), axis=1)

except np.exceptions.AxisError as err:
    print("AxisError:", err)
```

##### Example #2: Two dimensional array

```python
arr1 = np.arange(1, 17).reshape(4, 4)
arr1
```

```python
arr2 = np.arange(1, 17).reshape(4, 4)
arr2
```

Default value for `axis` parameter is `0`.

```python
np.concatenate((arr1, arr2))
```

When axis is explicitly set to zero.

```python
np.concatenate((arr1, arr2), axis=0)
```

When axis is explicitly set to one.

```python
np.concatenate((arr1, arr2), axis=1)
```

#### `np.vstack`

##### Example #1

```python
np.vstack((arr1, arr2))
```

##### Example #2

```python
arr = np.arange(1, 4)
arr
```

```python
op = np.vstack((arr, arr))

display(op)
print("\nShape:", op.shape)
```

```python
arr = np.arange(1, 5).reshape(2, 2)
arr
```

#### `np.hstack`

##### Example #1

```python
np.hstack((arr1, arr2))
```

##### Example #2

```python
arr = np.arange(1, 4)
arr
```

```python
op = np.hstack((arr, arr))

display(op)
print("\nShape:", op.shape)
```

```python
arr = np.arange(1, 5).reshape(2, 2)
arr
```

```python
op = np.hstack((arr, arr))

display(op)
print("\nShape:", op.shape)
```

Horizontally stacking two vertical arrays of same shape:

```python
a = np.array(
    [
        [1],
        [2],
        [3],
    ]
)
b = np.array(
    [
        [4],
        [5],
        [6],
    ]
)
np.hstack((a, b))
```

Horizontally stacking two vertical arrays of different shape:

```python
try:
    a = np.array(
        [
            [1],
            [2],
            [3],
        ]
    )
    b = np.array(
        [
            [4],
            [5],
            [6],
            [7],
        ]
    )
    np.hstack((a, b))

except ValueError as err:
    print("ValueError:", err)
```
