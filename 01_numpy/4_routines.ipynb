## 6 Sorting

```python
arr = np.array([10, 6, 0, 5, 7, 1])
arr
```

```python
arr.sort()
arr
```

> **Note**:
>
> NumPy uses Tim sort algorithm to sort elements in the array

#### Axis in Sorting

> **Note**:
>
> Default value for axis in `np.sort` is -1, hence NumPy always sorts array by last dimension by default

### 6.1 Sorting on 1D Array

```python
arr = np.array([16, 2, 13, 9])
arr
```

##### No Axis

```python
np.sort(arr)
```

##### Axis Zero

```python
np.sort(arr, axis=0)
```

##### Axis One

```python
try:
    np.sort(arr, axis=1)

except np.exceptions.AxisError as err:
    print("AxisError:", err)
```

### 6.2 Sorting 2D Array

```python
arr = np.array(
    [
        [16, 2, 13, 9],
        [1, 6, 17, 18],
        [4, 10, 7, 11],
        [13, 18, 5, 6],
    ]
)
arr
```

##### No Axis

```python
np.sort(arr)
```

##### Axis Zero

```python
np.sort(arr, axis=0)
```

##### Axis One

```python
np.sort(arr, axis=1)
```

### 6.3 Sorting 3D Array

```python
arr = np.array(
    [
        [
            [16, 2, 9],
            [1, 6, 18],
            [4, 7, 11],
            [3, 18, 5],
        ],
        [
            [1, 4, 13],
            [6, 2, 10],
            [13, 5, 7],
            [6, 11, 9],
        ],
    ]
)
print(arr)
print("\nShape:", arr.shape)
```

##### No Axis

```python
np.sort(arr)
```

##### Axis Zero

```python
np.sort(arr, axis=0)
```

##### Axis One

```python
np.sort(arr, axis=1)
```

##### Axis Two

```python
np.sort(arr, axis=2)
```

##### Axis Three

```python
try:
    np.sort(arr, axis=3)

except np.exceptions.AxisError as err:
    print(err)
```

## 7 Random Sampling

### `rand`

```python
np.random.rand(5)
```

```python
arr = np.random.rand(2, 4, 3)

display(arr)
arr.shape
```

### `randn`

```python
np.random.randn(5)
```

```python
arr = np.random.randn(2, 4, 3)

display(arr)
arr.shape
```

### `randint`

```python
np.random.randint(low=0, high=256, size=5, dtype="uint8")
```

```python
arr = np.random.randint(2, size=(2, 4, 3))

display(arr)
arr.shape
```

### `random`

```python
np.random.random(5)
```

```python
arr = np.random.random(size=(3, 2, 4))

display(arr)
arr.shape
```

### `choice`

```python
np.random.choice(a=["Red", "Red", "Red", "Green", "Green"], size=3, replace=True)
```

```python
# By default `replace` is set to True.
np.random.choice(a=["Red", "Red", "Red", "Green", "Green"], size=4, replace=False)
```
