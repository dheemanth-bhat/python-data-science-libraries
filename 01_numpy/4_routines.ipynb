#### `where` without `x` and `y`

Passing an NumPy array to `np.where` without `x` and `y` has two cases:

1. NumPy array without condition returns indexes of non-zero elements.
2. NumPy array with condition returns indexes of elements matching the condition.

```python
arr_1d = np.array([6, -4, 0, 8, 5, 3, 10, 2])
arr_1d
```

```python
arr_2d = np.array(
    [
        [0, 1, 2],
        [3, 0, 4],
        [5, 6, 0],
    ]
)
arr_2d
```

##### Case #1: Pass array without condition

```python
np.where(arr_1d)
```

```python
np.where(arr_2d)
```

First array is row indices, second array is column indices.

##### Case #2: Pass array with condition

```python
np.where(arr_1d > 4)
```

```python
np.where(arr_2d == 0)
```

First array is row indices, second array is column indices.

## 6 Sorting

```python
arr = np.array([10, 6, 0, 5, 7, 1])
arr
```

```python
arr.sort()
arr
```

> **Note**:
>
> NumPy uses Tim sort algorithm to sort elements in the array

#### Axis in Sorting

> **Note**:
>
> Default value for axis in `np.sort` is -1, hence NumPy always sorts array by last dimension by default

### 6.1 Sorting on 1D Array

```python
arr = np.array([16, 2, 13, 9])
arr
```

##### No Axis

```python
np.sort(arr)
```

##### Axis Zero

```python
np.sort(arr, axis=0)
```

##### Axis One

```python
try:
    np.sort(arr, axis=1)

except np.exceptions.AxisError as err:
    print("AxisError:", err)
```

### 6.2 Sorting 2D Array

```python
arr = np.array(
    [
        [16, 2, 13, 9],
        [1, 6, 17, 18],
        [4, 10, 7, 11],
        [13, 18, 5, 6],
    ]
)
arr
```

##### No Axis

```python
np.sort(arr)
```

##### Axis Zero

```python
np.sort(arr, axis=0)
```

##### Axis One

```python
np.sort(arr, axis=1)
```

### 6.3 Sorting 3D Array

```python
arr = np.array(
    [
        [
            [16, 2, 9],
            [1, 6, 18],
            [4, 7, 11],
            [3, 18, 5],
        ],
        [
            [1, 4, 13],
            [6, 2, 10],
            [13, 5, 7],
            [6, 11, 9],
        ],
    ]
)
print(arr)
print("\nShape:", arr.shape)
```

##### No Axis

```python
np.sort(arr)
```

##### Axis Zero

```python
np.sort(arr, axis=0)
```

##### Axis One

```python
np.sort(arr, axis=1)
```

##### Axis Two

```python
np.sort(arr, axis=2)
```

##### Axis Three

```python
try:
    np.sort(arr, axis=3)

except np.exceptions.AxisError as err:
    print(err)
```

## 7 Random Sampling

### `rand`

`np.random.rand()` generates ransom sample from **Uniform distribution**.

```python
np.random.rand(5)
```

```python
x = np.random.rand(1000)

plt.figure(figsize=(5, 3))
sns.histplot(x=x)
plt.title("Uniform Distribution")

plt.show()
```

```python
arr = np.random.rand(2, 4, 3)

display(arr)
arr.shape
```

### `randn`

`np.random.randn()` generates ransom sample from **Standard Normal (Gaussian) distribution**.

```python
np.random.randn(5)
```

```python
x = np.random.randn(1000)

plt.figure(figsize=(5, 3))
sns.histplot(x=x)
plt.title("Standard Normal Distribution")

plt.show()
```

```python
arr = np.random.randn(2, 4, 3)

display(arr)
arr.shape
```

### `randint`

```python
np.random.randint(low=0, high=256, size=5, dtype="uint8")
```

```python
x = np.random.randint(low=0, high=256, size=1000, dtype="uint8")

plt.figure(figsize=(5, 3))
sns.histplot(x=x)
plt.title("Standard Normal Distribution")

plt.show()
```

```python
arr = np.random.randint(2, size=(2, 4, 3))

display(arr)
arr.shape
```
